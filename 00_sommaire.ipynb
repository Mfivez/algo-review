# Préparation algorithmique — comprendre avant d'appliquer

Cette série de notebooks prépare aux tests techniques d'algorithmique en temps limité.
L'objectif n'est pas d'apprendre des recettes par cœur, mais de **comprendre chaque outil** : ce qu'il est, comment il fonctionne, pourquoi il est rapide, et seulement ensuite **quand l'utiliser**.

Chaque notebook est **indépendant** : il contient tout ce qu'il faut pour être exécuté seul. Suis quand même l'ordre : chaque thème réutilise les idées des précédents.

## Comment est construit chaque thème
1. **Le problème de départ** — un exercice concret qui donne une raison d'apprendre le concept.
2. **Le concept expliqué** — ce que c'est, comment ça marche, avec une image mentale.
3. **On manipule** — de petites cellules de code à exécuter et à modifier.
4. **On déroule à la main** — une trace pas à pas de l'algorithme, étape par étape.
5. **Quand y penser ?** — les mots de l'énoncé qui doivent faire penser à cet outil.
6. **Les pièges** — les erreurs que font presque tous les débutants.
7. **À toi de jouer** — un exercice avec tests automatiques, et la solution cachée.

> **Conseil** : exécute les cellules dans l'ordre (`Shift + Entrée`). Change les exemples, casse le code exprès : c'est en voyant ce qui ne marche pas qu'on comprend pourquoi le code est écrit comme il l'est.
